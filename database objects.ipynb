```sql
create function databricks_practice.bronze.get_fullName_ufd(first_name string, last_name string)
returns string
return first_name || '' || last_name
```

```sql
select customer_id, get_fullName_ufd(first_name, last_name) as full_name, review from databricks_practice.bronze.customer_reviews;
```
